```python
file_types = {
    "txt": "Text File",
    ".pdf": "PDF Document",
    "jpg": "JPEG Image",
}
# use of in and not in orerators

print(".pdf" in file_types)
print(".mp3" in file_types)
print(".mp3" not in file_types)
```
