```python
a = int(input("Enter value of a "))
b = int(input("Enter value of b "))
c = int(input("Enter value of c "))
if a>b and a>c:
  print("A is the largest number")
elif a<b and c<b:
  print("B is the largest number")
else:
  print("C is the largest number")

```
