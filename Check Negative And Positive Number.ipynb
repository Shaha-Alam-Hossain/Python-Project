```python
#Write a program to ask the user to enter a number and
#determine whether it is positive, negative, or zero.
```

```python
num = int(input("Enter a Number: "))
if num > 0:
  print("It is a positive number:",num)
elif num < 0:
  print("It is a Negative Number:",num)
else:
    print("It is a Zero")
```
