```python
#Write a program to ask the user to enter a number and
#determine whether it is positive, negative, or zero.
```

```python
num = int(input("Enter a Number: "))
if num > 0:
  print("It is a positive number:",num)
elif num < 0:
  print("It is a Negative Number:",num)
else:
    print("It is a Zero")
```

```python
#Write a program that asks the user to enter three numbers and
#determines which one is the largest.

a = int(input("Enter The Value of A: "))
b = int(input("Enter The Value of B: "))
c = int(input("Enter The Value of C: "))

if a>b and a>c:
  print("A is the largest Number ")
elif b>a and b>c:
  print("B is the largest Number ")
else:
  print("C is the largest Number ")
```
